```python
import numpy as np
import time
nums=list(range(1000000))
arr=np.arange(1000000)
t0=time.time()
sum(nums)
print("Time taken by list : ",time.time()-t0)
t0=time.time()
arr.sum()
print("Time taken by array : ",time.time()-t0)
```

```python
#Numpy(Creating arrays)
a=np.array([2,4,6,8])
b=np.zeros(5)
c=np.ones((2,3))
d=np.arange(0,10,2)
e=np.linspace(0,1,5)
```

```python
print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
f=np.arange(5,51,5)
print(f)
```

```python
arr_reshape=f.reshape((2,5))
print(arr_reshape)
```

```python
print(arr_reshape.shape)
print(arr_reshape.ndim)
print(arr_reshape.size)
print(arr_reshape.dtype)
```

```python
#Indexing and slicing
a=np.array([10,20,30,40,50])
print(a[0])
print(a[-1])
print(a[1:4])
print(a[:3])
print(a[::2])
```

```python
m=np.array([[1,2,3],[4,5,6]])
print(m[0,2])
print(m[:,1])
print(m[1,:])
```
